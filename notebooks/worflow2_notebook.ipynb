```python
# ============================================
# WORKFLOW 2: Ribosome + EF-G + S12 alignment Runner (general)
# ============================================

import sys
import os

# Add src folder to path
sys.path.append(r"C:\Protein Modelling GitHub Project\src")

from ribosome_drug_proximity import run_workflow2

# --- Folders ---
input_folder = r"C:\Protein Modelling GitHub Project\data"
output_folder = r"C:\Protein Modelling GitHub Project\models"
os.makedirs(output_folder, exist_ok=True)

# --- List all CIF files in input folder ---
cif_files = [f for f in os.listdir(input_folder) if f.endswith(".cif")]

# --- Pair CIF files automatically ---
# Example: assume first file is ribosome (target), others are EF-G (to superimpose)
# You can adjust logic as needed
for target_file in cif_files:
    for query_file in cif_files:
        if target_file == query_file:
            continue  # skip pairing file with itself

        cif_4v53 = os.path.join(input_folder, target_file)
        cif_4v5f = os.path.join(input_folder, query_file)

        # Create output filename dynamically
        target_base = os.path.splitext(target_file)[0]
        query_base = os.path.splitext(query_file)[0]
        output_file = os.path.join(output_folder, f"{query_base}_on_{target_base}.cif")

        # Run workflow
        run_workflow2(
            cif_4v5f=cif_4v5f,
            cif_4v53=cif_4v53,
            output_file=output_file,
            visualize=False  # Set True if you want visualization
        )

print("All CIF file pairs processed. Outputs saved in:", output_folder)
```
